```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

url = "https://en.wikipedia.org/wiki/List_of_countries_and_dependencies_by_population"

headers = {"User-Agent": "Mozilla/5.0"}
req = requests.get(url, headers=headers)

soup = BeautifulSoup(req.content, "html.parser")

# get all wiki tables
tables = soup.find_all("table", {"class": "wikitable"})

# select the first main table
table = tables[0]

rows = table.find_all("tr")

data = []

for row in rows[1:]:
    cells = row.find_all("td")
    
    if len(cells) >= 3:
        rank = cells[0].text.strip()
        country = cells[1].text.strip()
        population = cells[2].text.strip()

        data.append([rank, country, population])

df = pd.DataFrame(data, columns=["Rank", "Country", "Population"])
print(df)
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

url = "https://en.wikipedia.org/wiki/List_of_countries_and_dependencies_by_population"

headers = {"User-Agent": "Mozilla/5.0"}
req = requests.get(url, headers=headers)

soup = BeautifulSoup(req.content, 'html.parser')

# ✅ FIXED LINE
table = soup.find_all("table", {"class": "wikitable"})[0]

rows = table.find_all('tr')

data = pd.DataFrame(columns=['Rank', 'Country', 'Population'])

for row in rows:
    cells = row.find_all('td')
    
    if len(cells) > 2:
        rank = cells[0].text.strip()
        country_cell = cells[1].find('a')
        
        if country_cell:
            country = country_cell.text.strip()
        else:
            country = cells[1].text.strip()

        population = cells[2].text.strip()

        data = pd.concat([
            data,
            pd.DataFrame({'Rank': [rank], 'Country': [country], 'Population': [population]})
        ], ignore_index=True)

data
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

url = 'https://en.wikipedia.org/wiki/World_population'

response = requests.get(url)
soup = BeautifulSoup(response.text, 'html.parser')

# Find all the tables on the webpage
tables = soup.find_all('table')

# For each table, create a DataFrame and print it
for i, table in enumerate(tables):
    # The [1:] is to skip the header row
    rows = table.find_all('tr')[1:]
    data = []
    for row in rows:
        cols = row.find_all('td')
        cols = [col.text.strip() for col in cols]
        data.append(cols)
    df = pd.DataFrame(data)
    print(f"Table {i}:")
    print(df)
    print("\n\n")
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

url = "https://en.wikipedia.org/wiki/World_population"

headers = {"User-Agent": "Mozilla/5.0"}
response = requests.get(url, headers=headers)

soup = BeautifulSoup(response.text, "html.parser")

# only wiki tables (better than all tables)
tables = soup.find_all("table", {"class": "wikitable"})

dataframes = []

for table in tables:
    rows = table.find_all("tr")
    data = []

    for row in rows:
        cols = row.find_all(["td", "th"])
        cols = [col.text.strip() for col in cols]

        if cols:  # skip empty rows
            data.append(cols)

    df = pd.DataFrame(data)
    dataframes.append(df)

# ✅ SAFE OUTPUT (no IndexError)
print("Total tables found:", len(dataframes))

# show all tables sizes
for i, df in enumerate(dataframes):
    print(i, df.shape)

# show first useful table
dataframes[0].head()
```
